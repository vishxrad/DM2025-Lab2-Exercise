## 1. Model Development (10 pts Required)

**Describe briefly each section, you can add graphs/charts to support your explanations.**

### 1.1 Preprocessing Steps

*   **Data Loading & Merging:** Loaded the raw JSON data (`final_posts.json`) and merged it with the split identification and emotion labels.
*   **Handling Missing Values:** Filled any `NaN` values in the text column with empty strings to prevent processing errors.
*   **Demojization:** Used the `emoji` library to convert emojis into text descriptions (e.g., `😂` to `:face_with_tears_of_joy:`). This is crucial for emotion recognition as emojis carry significant sentiment weight that would otherwise be lost or treated as noise.

### 1.2 Feature Engineering Steps

I employed a hybrid feature extraction strategy combining statistical text vectors with lexicon-based and structural features:

1.  **TF-IDF Vectorization:**
    *   **Word-level:** Extracted 1-gram to 3-grams (max 30k features) to capture phrases and context.
    *   **Character-level:** Extracted 2-gram to 5-grams (max 15k features) to capture subword information, misspellings, and morphological variations.
2.  **Sentiment & Structural Features:**
    *   **VADER Sentiment:** Calculated `pos`, `neg`, `neu`, and `compound` scores using the VADER lexicon.
    *   **Structural:** Counted occurrences of `!` and `?`, calculated the ratio of uppercase letters (to detect "shouting"), and text length.
    *   These extra features were scaled using `MinMaxScaler`.
3.  **Feature Stacking & Selection:**
    *   Concatenated all features (Word TF-IDF + Char TF-IDF + Extra Features) into a single sparse matrix.
    *   Applied **Chi-Squared (`SelectKBest`)** to select the top 20,000 most informative features, reducing noise and dimensionality.

### 1.3 Explanation of Your Model

My final model is a **Soft Voting Ensemble** enhanced with **Pseudo-Labeling**:

1.  **Base Models:**
    *   **Linear SVC:** Hyperparameter tuned (`C` found via GridSearchCV). Wrapped in `CalibratedClassifierCV` (sigmoid) to output probabilities.
    *   **Logistic Regression:** A robust linear baseline.
2.  **Ensemble:** A `VotingClassifier` with `voting='soft'` averages the predicted probabilities of the SVM and LR models.
3.  **Pseudo-Labeling (Semi-Supervised Learning):**
    *   Trained the ensemble on the original training set.
    *   Predicted probabilities for the **test set**.
    *   Selected test samples where the model was highly confident (probability > 0.95).
    *   Added these "pseudo-labeled" test samples to the training data and **retrained** the model. This allows the model to learn from the distribution of the test data.

# **Competition Code**

## 1. Preprocessing Steps

```python
import pandas as pd
import json
import emoji
import numpy as np
from scipy.sparse import hstack, csr_matrix
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import VotingClassifier
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

print("Loading data...")
with open('dm-lab-2-private-competition/final_posts.json', 'r') as f:
    data = json.load(f)

posts = [{'id': x['root']['_source']['post']['post_id'], 
          'text': x['root']['_source']['post']['text']} for x in data]
df_text = pd.DataFrame(posts)
df_split = pd.read_csv('dm-lab-2-private-competition/data_identification.csv')
df_labels = pd.read_csv('dm-lab-2-private-competition/emotion.csv')

df_full = df_text.merge(df_split, on='id').merge(df_labels, on='id', how='left')
df_full['text'] = df_full['text'].fillna('').astype(str)
df_full['clean_text'] = df_full['text'].apply(lambda x: emoji.demojize(x, delimiters=(" ", " ")))

print("Data loaded and preprocessed.")
df_full.tail()
```

## 2. Feature Engineering Steps

```python
analyzer = SentimentIntensityAnalyzer()
def get_extra_features(text):
    scores = analyzer.polarity_scores(text)
    return [
        scores['neg'], scores['neu'], scores['pos'], scores['compound'],
        text.count('!'), text.count('?'), 
        sum(1 for c in text if c.isupper()) / (len(text) + 1),
        len(text)
    ]

print("Extracting extra features...")
extra_features = np.array([get_extra_features(t) for t in df_full['text']])
scaler = MinMaxScaler()
extra_features_scaled = scaler.fit_transform(extra_features)

print("Vectorizing text...")
word_vect = TfidfVectorizer(analyzer='word', token_pattern=r'(?u)\b\w\w+\b', 
                            ngram_range=(1, 3), max_features=30000, sublinear_tf=True) # Increased n-grams
X_word = word_vect.fit_transform(df_full['clean_text'])

char_vect = TfidfVectorizer(analyzer='char_wb', ngram_range=(2, 5), 
                            max_features=15000, sublinear_tf=True)
X_char = char_vect.fit_transform(df_full['clean_text'])

# Stack
X_full_raw = hstack([X_word, X_char, csr_matrix(extra_features_scaled)])

train_mask = df_full['split'] == 'train'
test_mask = df_full['split'] == 'test'
y_train = LabelEncoder().fit_transform(df_full[train_mask]['emotion'])

print("Selecting best features...")
selector = SelectKBest(chi2, k=20000) 
X_train_selected = selector.fit_transform(X_full_raw[train_mask], y_train)
X_test_selected = selector.transform(X_full_raw[test_mask])
print(f"Feature engineering complete. Training shape: {X_train_selected.shape}")
```

## 3. Model Implementation Steps

```python
# 1. Hyperparameter Tuning
print("Tuning SVM...")
svm_param_grid = {'C': [0.1, 0.5, 1, 2, 5, 10]}
svm_search = GridSearchCV(LinearSVC(class_weight='balanced', dual=False, max_iter=5000), 
                          svm_param_grid, cv=5, n_jobs=-1, verbose=1)
svm_search.fit(X_train_selected, y_train)
best_C = svm_search.best_params_['C']
print(f"Found Best SVM C: {best_C}")

# 2. Model Definition
clf1 = LinearSVC(C=best_C, class_weight='balanced', dual=False, max_iter=10000)
from sklearn.calibration import CalibratedClassifierCV
clf1_calibrated = CalibratedClassifierCV(clf1, method='sigmoid') # Calibrate for voting

clf2 = LogisticRegression(C=2.0, solver='liblinear', class_weight='balanced')

model = VotingClassifier(estimators=[('svm', clf1_calibrated), ('lr', clf2)], voting='soft', n_jobs=-1)

# 3. Initial Training & Pseudo-Labeling
print("Training initial model...")
model.fit(X_train_selected, y_train)
probs = model.predict_proba(X_test_selected) # Get probabilities for test data

# Find confident predictions (e.g., > 95% sure)
confidence_threshold = 0.95
confident_indices = np.where(np.max(probs, axis=1) > confidence_threshold)[0]
confident_labels = np.argmax(probs[confident_indices], axis=1)

print(f"Found {len(confident_indices)} confident test samples to add to training.")

X_pseudo_train = hstack([X_train_selected.T, X_test_selected[confident_indices].T]).T
y_pseudo_train = np.concatenate([y_train, confident_labels])

# 4. Retraining with Pseudo-Labels
print("Retraining with pseudo-labels...")
model.fit(X_pseudo_train, y_pseudo_train)

# 5. Final Prediction
final_preds = model.predict(X_test_selected)
le = LabelEncoder().fit(df_full[train_mask]['emotion'])
final_labels = le.inverse_transform(final_preds)

submission = pd.DataFrame({'id': df_full[test_mask]['id'], 'emotion': final_labels})
submission.to_csv('submission.csv', index=False)
print("Submission saved to submission.csv")
```

```python
from sklearn.base import clone
from sklearn.model_selection import cross_validate, train_test_split
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# 1. Cross-Validation for robust metrics
print("--- Cross Validation Results (5-Fold) ---")
scoring = ['accuracy', 'f1_macro', 'recall_macro', 'precision_macro']
# We use a clone to not disturb the fitted model
cv_results = cross_validate(clone(model), X_train_selected, y_train, cv=5, scoring=scoring, n_jobs=-1)

print(f"Mean Accuracy: {cv_results['test_accuracy'].mean():.4f}")
print(f"Mean F1 Macro: {cv_results['test_f1_macro'].mean():.4f}")
print(f"Mean Recall Macro: {cv_results['test_recall_macro'].mean():.4f}")
print(f"Mean Precision Macro: {cv_results['test_precision_macro'].mean():.4f}")

# 2. Detailed Classification Report on a hold-out set
print("\n--- Detailed Classification Report (Validation Set) ---")
X_tr, X_val, y_tr, y_val = train_test_split(X_train_selected, y_train, test_size=0.2, random_state=42, stratify=y_train)
val_model = clone(model)
val_model.fit(X_tr, y_tr)
y_pred_val = val_model.predict(X_val)

# Ensure LabelEncoder is available for class names
if 'le' not in locals():
    le = LabelEncoder().fit(df_full[train_mask]['emotion'])

print(classification_report(y_val, y_pred_val, target_names=le.classes_))

# 3. Confusion Matrix
print("\n--- Confusion Matrix ---")
cm = confusion_matrix(y_val, y_pred_val)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=le.classes_)
fig, ax = plt.subplots(figsize=(10, 10))
disp.plot(cmap='Blues', ax=ax, xticks_rotation='vertical')
plt.title("Confusion Matrix (Validation Set)")
plt.show()
```
